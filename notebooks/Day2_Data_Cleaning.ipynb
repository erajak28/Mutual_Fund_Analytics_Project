```python
import pandas as pd
import os

base_path = r"C:\Users\WINDOWS\OneDrive\Documents\Mutual_Fund_Analytics_Project"
raw_path = base_path + r"\data\raw"
processed_path = base_path + r"\data\processed"

os.makedirs(processed_path, exist_ok=True)
```

```python
fund_master = pd.read_csv(raw_path + r"\01_fund_master.csv")

print(fund_master.shape)
print(fund_master.isnull().sum())
print("Duplicates:", fund_master.duplicated().sum())

fund_master["launch_date"] = pd.to_datetime(fund_master["launch_date"])

print("Invalid expense ratio:", ((fund_master["expense_ratio_pct"] < 0) | (fund_master["expense_ratio_pct"] > 3)).sum())
print("Invalid SIP amount:", (fund_master["min_sip_amount"] <= 0).sum())
print("Duplicate AMFI codes:", fund_master["amfi_code"].duplicated().sum())

fund_master = fund_master.drop_duplicates()

fund_master.to_csv(processed_path + r"\clean_fund_master.csv", index=False)
```

```python
aum = pd.read_csv(raw_path + r"\03_aum_by_fund_house.csv")

print(aum.shape)
print(aum.isnull().sum())
print("Duplicates:", aum.duplicated().sum())

aum["date"] = pd.to_datetime(aum["date"])

print("Invalid AUM:", (aum["aum_crore"] <= 0).sum())
print("Invalid scheme count:", (aum["num_schemes"] <= 0).sum())

aum = aum.drop_duplicates()

aum.to_csv(processed_path + r"\clean_aum_by_fund_house.csv", index=False)
```

```python
sip = pd.read_csv(raw_path + r"\04_monthly_sip_inflows.csv")

print(sip.shape)
print(sip.isnull().sum())
print("Duplicates:", sip.duplicated().sum())

sip["month"] = pd.to_datetime(sip["month"])

print("Invalid SIP inflow:", (sip["sip_inflow_crore"] <= 0).sum())
print("Invalid active SIP accounts:", (sip["active_sip_accounts_crore"] <= 0).sum())

sip = sip.drop_duplicates()

sip.to_csv(processed_path + r"\clean_monthly_sip_inflows.csv", index=False)
```

```python
category_inflows = pd.read_csv(raw_path + r"\05_category_inflows.csv")

print(category_inflows.shape)
print(category_inflows.isnull().sum())
print("Duplicates:", category_inflows.duplicated().sum())

category_inflows["month"] = pd.to_datetime(category_inflows["month"])

print("Invalid category inflows:", category_inflows["net_inflow_crore"].isnull().sum())

category_inflows = category_inflows.drop_duplicates()

category_inflows.to_csv(processed_path + r"\clean_category_inflows.csv", index=False)
```

```python
folio = pd.read_csv(raw_path + r"\06_industry_folio_count.csv")

print(folio.shape)
print(folio.isnull().sum())
print("Duplicates:", folio.duplicated().sum())

folio["month"] = pd.to_datetime(folio["month"])

print("Invalid total folios:", (folio["total_folios_crore"] <= 0).sum())

folio = folio.drop_duplicates()

folio.to_csv(processed_path + r"\clean_industry_folio_count.csv", index=False)
```

```python
holdings = pd.read_csv(raw_path + r"\09_portfolio_holdings.csv")

print(holdings.shape)
print(holdings.isnull().sum())
print("Duplicates:", holdings.duplicated().sum())

holdings["portfolio_date"] = pd.to_datetime(holdings["portfolio_date"])

print("Invalid weight %:", ((holdings["weight_pct"] < 0) | (holdings["weight_pct"] > 100)).sum())
print("Invalid market value:", (holdings["market_value_cr"] <= 0).sum())
print("Invalid stock price:", (holdings["current_price_inr"] <= 0).sum())

holdings = holdings.drop_duplicates()

holdings.to_csv(processed_path + r"\clean_portfolio_holdings.csv", index=False)
```

```python
benchmark = pd.read_csv(raw_path + r"\10_benchmark_indices.csv")

print(benchmark.shape)
print(benchmark.isnull().sum())
print("Duplicates:", benchmark.duplicated().sum())

benchmark["date"] = pd.to_datetime(benchmark["date"])

print("Invalid close value:", (benchmark["close_value"] <= 0).sum())

benchmark = benchmark.drop_duplicates()

benchmark.to_csv(processed_path + r"\clean_benchmark_indices.csv", index=False)
```
