```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import os

base_path = r"C:\Users\WINDOWS\OneDrive\Documents\Mutual_Fund_Analytics_Project"

processed_path = base_path + r"\data\processed"
charts_path = base_path + r"\reports\charts"

os.makedirs(charts_path, exist_ok=True)

print("Setup completed")
```

```python
#loading dataset

fund_master = pd.read_csv(processed_path + r"\clean_fund_master.csv")
nav = pd.read_csv(processed_path + r"\clean_nav.csv")
aum = pd.read_csv(processed_path + r"\clean_aum_by_fund_house.csv")
sip = pd.read_csv(processed_path + r"\clean_monthly_sip_inflows.csv")
category_inflows = pd.read_csv(processed_path + r"\clean_category_inflows.csv")
transactions = pd.read_csv(processed_path + r"\clean_transactions.csv")
folio = pd.read_csv(processed_path + r"\clean_industry_folio_count.csv")
holdings = pd.read_csv(processed_path + r"\clean_portfolio_holdings.csv")
benchmark = pd.read_csv(processed_path + r"\clean_benchmark_indices.csv")

print("All datasets loaded successfully")
```

```python
nav["date"] = pd.to_datetime(nav["date"])

fig = px.line(
    nav,
    x="date",
    y="nav",
    color="amfi_code",
    title="Daily NAV Trend for All 40 Mutual Fund Schemes (2022–2026)"
)

fig.add_vrect(
    x0="2023-01-01",
    x1="2023-12-31",
    fillcolor="green",
    opacity=0.15,
    layer="below",
    line_width=0,
    annotation_text="2023 Bull Run",
    annotation_position="top left"
)

fig.add_vrect(
    x0="2024-06-01",
    x1="2024-10-31",
    fillcolor="red",
    opacity=0.15,
    layer="below",
    line_width=0,
    annotation_text="2024 Market Correction",
    annotation_position="top left"
)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="NAV",
    legend_title="AMFI Code",
    height=700,
    width=1100
)

fig.write_image(
    charts_path + r"\nav_trend_all_40_schemes.png"
)
fig.show()
```

### Insight 1: Overall NAV Growth

Most mutual fund schemes exhibited a positive NAV trend between 2022 and 2026, indicating long-term market growth and wealth creation.

### Insight 2: Strong 2023 Rally

The majority of schemes experienced accelerated growth during 2023, reflecting favorable market conditions and strong investor sentiment.

### Insight 3: Temporary 2024 Correction

Several schemes showed slower growth and temporary declines during 2024, indicating a period of market correction before resuming their upward trend.


### Insight 4: Presence of High-NAV Outliers

A few schemes maintained significantly higher NAV values than the rest of the portfolio, indicating substantial differences in scheme maturity and historical performance.

```python
avg_nav = nav.groupby("date")["nav"].mean().reset_index()

avg_nav.head()
```

```python
plt.figure(figsize=(14,6))

plt.plot(avg_nav["date"],
         avg_nav["nav"])

plt.title("Average NAV Trend (2022-2025)")
plt.xlabel("Date")
plt.ylabel("Average NAV")

plt.grid(True)

plt.savefig(charts_path + r"\nav_trend_average.png",
            bbox_inches="tight")

plt.show()
```

### Insight:
Although individual scheme NAVs vary significantly, the average NAV trend shows consistent growth from 2022 to 2026. This indicates strong long-term market performance across the mutual fund industry.

```python
# Find all-time high SIP inflow
max_sip = sip.loc[sip["sip_inflow_crore"].idxmax()]

# Create Plotly line chart
fig = px.line(
    sip,
    x="month",
    y="sip_inflow_crore",
    markers=True,
    title="Monthly SIP Inflow Trend (Jan 2022 - Dec 2025)"
)

# Highlight all-time high point
fig.add_annotation(
    x=max_sip["month"].strftime("%Y-%m-%d"),
    y=max_sip["sip_inflow_crore"],
    text=f"All-Time High<br>₹{max_sip['sip_inflow_crore']:,} Cr<br>Dec 2025",
    showarrow=True,
    arrowhead=2,
    ax=-100,
    ay=-80
)

# Format chart
fig.update_layout(
    xaxis_title="Month",
    yaxis_title="SIP Inflow (₹ Crore)",
    height=600,
    width=1100,
    template="plotly_white"
)

# Display chart
fig.show()

# Save chart
fig.write_image(charts_path + r"\sip_inflow_trend.png")

print("Chart saved successfully!")
```

### Insight: Record SIP Inflows in December 2025

Monthly SIP inflows showed a strong upward trend from approximately ₹11,500 crore in January 2022 to an all-time high of ₹31,002 crore in December 2025.

The consistent increase in SIP contributions indicates growing retail investor participation, rising financial awareness, and wider adoption of disciplined long-term investing through mutual funds.

```python
pivot_data = category_inflows.pivot_table(
    index="category",
    columns="month",
    values="net_inflow_crore"
)
```

```python
pivot_data.head()
```

```python
plt.figure(figsize=(14,6))

sns.heatmap(
    pivot_data,
    cmap="YlOrRd",
    linewidths=0.5
)

plt.title("Category-wise Net Inflow Heatmap (2024-2025)")
plt.xlabel("Month")
plt.ylabel("Fund Category")

plt.tight_layout()

plt.savefig(
    charts_path + r"\category_inflow_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight : Category-wise Investor Preferences

The heatmap reveals that Liquid Funds consistently received the highest net inflows across most months, indicating strong investor preference for liquidity and short-term fund parking.

Among equity-oriented categories, Flexi Cap and Large & Mid Cap funds attracted relatively higher inflows, reflecting investor interest in diversified growth opportunities.

Debt-oriented categories such as Gilt and Short Duration funds recorded comparatively lower inflows, suggesting that investors were more focused on growth-oriented investment options during the period.

```python
# Age group distribution pie chart

age_counts = transactions["age_group"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    age_counts,
    labels=age_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Investor Age Group Distribution")
plt.tight_layout()

plt.savefig(charts_path + r"\age_group_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Insight:

The 26–35 age group represents the largest investor segment, accounting for 41.1% of all investors. This indicates that young working professionals are the most active participants in mutual fund investments.

The 36–45 age group contributes 24.9%, showing strong participation from mid-career investors focused on long-term wealth creation.

Investors aged 56+ account for only 7.5%, suggesting relatively lower participation among senior investors.

```python
#SIP amount box plot by age group

sip_transactions = transactions[transactions["transaction_type"] == "SIP"]

plt.figure(figsize=(10,6))

sns.boxplot(
    data=sip_transactions,
    x="age_group",
    y="amount_inr"
)

plt.title("SIP Amount Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("SIP Amount (INR)")

plt.tight_layout()

plt.savefig(charts_path + r"\sip_amount_by_age_group.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Insight:

The box plot shows that SIP investment behavior is broadly similar across age groups, with median SIP amounts concentrated around similar levels.

However, all age groups contain several high-value SIP outliers, indicating the presence of investors making significantly larger contributions than the average investor.

The 56+ and 46–55 age groups show slightly higher upper ranges, suggesting that experienced investors may allocate larger amounts to systematic investments.

```python
#Gender split chart

gender_counts = transactions["gender"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Investor Gender Split")
plt.tight_layout()

plt.savefig(charts_path + r"\gender_split.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Insight:

Male investors account for 66.5% of the investor base, while female investors account for 33.5%.

```python
sip_state = transactions[transactions["transaction_type"] == "SIP"]

state_sip = sip_state.groupby("state")["amount_inr"].sum().sort_values(ascending=False)

state_sip.head(10)
```

```python
plt.figure(figsize=(10,6))

state_sip.head(10).sort_values().plot(
    kind="barh"
)

plt.title("Top 10 States by SIP Investment Amount")
plt.xlabel("Total SIP Amount (INR)")
plt.ylabel("State")

plt.tight_layout()

plt.savefig(
    charts_path + r"\sip_amount_by_state.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight :

The analysis shows that Madhya Pradesh recorded the highest SIP investment amount, followed by Punjab and Telangana.

Other major contributors include Gujarat, Tamil Nadu, Haryana, and Karnataka, indicating strong mutual fund participation across multiple regions of India.

The relatively close investment amounts among the top states suggest that mutual fund adoption is becoming geographically diversified rather than being concentrated in a few metropolitan regions.

```python
plt.figure(figsize=(14,6))

plt.plot(
    folio["month"],
    folio["total_folios_crore"],
    marker="o",
    linewidth=2
)

# First point
plt.annotate(
    "13.26 Cr",
    xy=(folio["month"].iloc[0], folio["total_folios_crore"].iloc[0]),
    xytext=(10,10),
    textcoords="offset points"
)

# Last point
plt.annotate(
    "26.12 Cr",
    xy=(folio["month"].iloc[-1], folio["total_folios_crore"].iloc[-1]),
    xytext=(10,10),
    textcoords="offset points"
)

plt.title("Folio Count Growth (2022-2025)")
plt.xlabel("Year")
plt.ylabel("Total Folios (Crore)")
plt.grid(True)

plt.tight_layout()

plt.savefig(
    charts_path + r"\folio_count_growth_milestones.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight: Strong Growth in Mutual Fund Folios

Total mutual fund folios increased from approximately 13.26 crore in early 2022 to 26.12 crore by the end of 2025.

The consistent upward trend indicates rising retail investor participation and increasing adoption of mutual funds as a preferred investment vehicle.

The near doubling of folio count during the period reflects growing financial awareness, improved digital access, and expanding mutual fund penetration across India.

```python
selected_funds = nav["amfi_code"].unique()[:10]
selected_funds
```

```python
nav_10 = nav[nav["amfi_code"].isin(selected_funds)]
```

```python
nav_pivot = nav_10.pivot_table(
    index="date",
    columns="amfi_code",
    values="nav")
```

```python
nav_pivot.head()
```

```python
daily_returns = nav_pivot.pct_change()
```

```python
daily_returns.head()
```

```python
corr_matrix = daily_returns.corr()

corr_matrix.head()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("NAV Return Correlation Matrix")

plt.tight_layout()

plt.savefig(
    charts_path + r"\nav_return_correlation_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight: Correlation Between Mutual Fund Returns

The correlation matrix shows generally weak relationships among the selected mutual funds, with most correlation values remaining close to zero.

This indicates that the funds exhibited relatively independent return patterns during the analysis period.

The low correlation levels suggest potential diversification benefits, as investing across these funds may help reduce portfolio concentration risk.

No strong positive or negative correlations were observed among the selected schemes.

```python
sector_weights = holdings.groupby("sector")["weight_pct"].sum().sort_values(ascending=False)

sector_weights
```

```python
plt.figure(figsize=(10,8))

wedges, texts, autotexts = plt.pie(
    sector_weights,
    labels=sector_weights.index,
    autopct="%1.1f%%",
    startangle=90
)

# Donut hole
centre_circle = plt.Circle((0,0), 0.70, fc="white")
fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Sector Allocation Across Equity Fund Portfolios")

plt.tight_layout()

plt.savefig(
    charts_path + r"\sector_allocation_donut.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Insight: Sector Allocation Across Equity Fund Portfolios

Banking emerged as the largest sector allocation, accounting for approximately 19.2% of total portfolio weight, indicating strong confidence in India's financial sector.

IT (13.4%) and Pharma (12.0%) were the next largest sectors, reflecting the importance of technology and healthcare in long-term investment strategies.

Automobile (9.5%) and Utilities (7.8%) also held significant allocations, providing diversification across cyclical and defensive sectors.

The overall sector distribution demonstrates a balanced investment approach, combining growth-oriented sectors such as IT and Automobile with relatively stable sectors such as Banking, Utilities, and FMCG.

1. Average NAV increased from approximately 208 in 2022 to 358 in 2026, indicating strong long-term growth across mutual fund schemes.
